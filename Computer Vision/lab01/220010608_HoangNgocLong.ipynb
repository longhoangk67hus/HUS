```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def load_image(path):
  return cv2.imread(path, cv2.IMREAD_GRAYSCALE)
```

```python
img = load_image('img.jpg')
plt.imshow(img, cmap='gray')
```

```python
def cal_histogram():
  hist = np.zeros(256)
  for i in range(img.shape[0]):
    for j in range(img.shape[1]):
      hist[img[i, j]] += 1
  return hist
```

```python
arr = cal_histogram()
def plot_histogram(histogram):
  plt.figure(figsize=(8, 6))
  plt.plot(histogram, color = 'black')
  plt.xlabel('Pixel Intensity (0 - 255)')
  plt.ylabel('Frequency')
  plt.xlim([0, 255])
  plt.title('Gray Image Histogram')
  plt.grid(True, linestyle='--', alpha=0.5)
  plt.show()
plot_histogram(arr)
```

```python
def grey_level_modification(img):
  modified_img = 16 * np.sqrt(img.astype(np.float32))
  modified_img = np.clip(modified_img, 0, 255).astype(np.uint8)
  return modified_img
mod_img = grey_level_modification(img)
plt.imshow(mod_img, cmap='gray')
```

```python
def Thresholding(histogram_img):
  threshold = np.mean(histogram_img)
  return threshold
threshold = Thresholding(arr)
print(threshold)
```
