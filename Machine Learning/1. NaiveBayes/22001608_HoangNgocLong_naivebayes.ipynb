```python
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# Select the target variable (y) and features (X)
X = df_orders.drop(columns=[col for col in df_orders.columns if col.startswith('cuisine_')])
y = df_orders[[col for col in df_orders.columns if col.startswith('cuisine_')]]

# Convert the one-hot encoded 'y' back to a single column of labels
y = y.idxmax(axis=1)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("Training features shape (X_train):", X_train.shape)
print("Testing features shape (X_test):", X_test.shape)
print("Training labels shape (y_train):", y_train.shape)
print("Testing labels shape (y_test):", y_test.shape)

# Train the Multinomial Naive Bayes model
model = MultinomialNB()
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred)

print("\nModel Evaluation:")
print(f"Accuracy: {accuracy:.4f}")
print("\nClassification Report:\n", report)
```

#### Bài 1

##### a. Hardcode

```python
# Merge training data with labels
train_data = train_values.merge(train_labels, on='building_id')

# Select only the required categorical features
categorical_features = [
    'land_surface_condition',
    'foundation_type', 
    'roof_type',
    'ground_floor_type',
    'other_floor_type',
    'position',
    'plan_configuration',
    'legal_ownership_status'
]

# Extract features and target
X_train = train_data[categorical_features]
y_train = train_data['damage_grade']
X_test = test_values[categorical_features]

print("Training data shape:", X_train.shape)
print("Test data shape:", X_test.shape)
print("Target variable distribution:")
print(y_train.value_counts().sort_index())
```

```python
def custom_multinomial_nb(X_train, y_train, X_test, alpha=1.0):
    """
    Custom implementation of Multinomial Naive Bayes
    """
    # Get unique classes and features
    classes = np.unique(y_train)
    features = X_train.columns
    
    # Calculate class probabilities P(y)
    class_probs = {}
    for cls in classes:
        class_probs[cls] = np.sum(y_train == cls) / len(y_train)
    
    # Calculate feature probabilities P(x_i|y) for each class
    feature_probs = {}
    
    for cls in classes:
        feature_probs[cls] = {}
        class_data = X_train[y_train == cls]
        
        for feature in features:
            feature_probs[cls][feature] = {}
            feature_values = np.unique(X_train[feature])
            
            for value in feature_values:
                # Count occurrences of this feature value in this class
                count = np.sum(class_data[feature] == value)
                # Total number of samples in this class
                total_class_samples = len(class_data)
                # Number of possible feature values
                num_feature_values = len(feature_values)
                
                # Apply Laplace smoothing
                prob = (count + alpha) / (total_class_samples + alpha * num_feature_values)
                feature_probs[cls][feature][value] = prob
    
    # Make predictions
    predictions = []
    
    for i in range(len(X_test)):
        class_scores = {}
        
        for cls in classes:
            # Start with log of class probability
            score = np.log(class_probs[cls])
            
            # Add log probabilities of each feature
            for feature in features:
                feature_value = X_test.iloc[i][feature]
                if feature_value in feature_probs[cls][feature]:
                    score += np.log(feature_probs[cls][feature][feature_value])
                else:
                    # Handle unseen feature values with smoothing
                    feature_values = list(feature_probs[cls][feature].keys())
                    num_feature_values = len(feature_values)
                    class_data = X_train[y_train == cls]
                    total_class_samples = len(class_data)
                    prob = alpha / (total_class_samples + alpha * (num_feature_values + 1))
                    score += np.log(prob)
            
            class_scores[cls] = score
        
        # Predict class with highest score
        predicted_class = max(class_scores.keys(), key=lambda x: class_scores[x])
        predictions.append(predicted_class)
    
    return np.array(predictions)

# Train and predict with custom implementation
print("Training Custom Multinomial Naive Bayes...")
custom_predictions = custom_multinomial_nb(X_train, y_train, X_test)

print(f"Custom implementation completed. Predictions shape: {custom_predictions.shape}")
print(f"Prediction distribution:")
unique_preds, counts = np.unique(custom_predictions, return_counts=True)
for pred, count in zip(unique_preds, counts):
    print(f"Class {pred}: {count} predictions")
```

```python
# 2. Trích xuất các trường nhị phân theo yêu cầu
print("Trích xuất 22 trường dữ liệu nhị phân...")

# Danh sách 22 trường nhị phân theo yêu cầu
binary_features = [
    # Superstructure features (i-xi)
    'has_superstructure_adobe_mud',           # i
    'has_superstructure_mud_mortar_stone',    # ii  
    'has_superstructure_stone_flag',          # iii
    'has_superstructure_cement_mortar_stone', # iv
    'has_superstructure_mud_mortar_brick',    # v
    'has_superstructure_cement_mortar_brick', # vi
    'has_superstructure_timber',              # vii
    'has_superstructure_bamboo',              # viii
    'has_superstructure_rc_non_engineered',   # ix
    'has_superstructure_rc_engineered',       # x
    'has_superstructure_other',               # xi
    
    # Secondary use features (xii-xxii)
    'has_secondary_use',                      # xii
    'has_secondary_use_agriculture',          # xiii
    'has_secondary_use_hotel',                # xiv
    'has_secondary_use_rental',               # xv
    'has_secondary_use_institution',          # xvi
    'has_secondary_use_school',               # xvii
    'has_secondary_use_industry',             # xviii
    'has_secondary_use_health_post',          # xix
    'has_secondary_use_gov_office',           # xx
    'has_secondary_use_use_police',           # xxi
    'has_secondary_use_other'                 # xxii
]

print(f"Số lượng trường nhị phân: {len(binary_features)}")

# Kiểm tra xem tất cả các trường có tồn tại trong dữ liệu không
missing_features = [feat for feat in binary_features if feat not in train_values.columns]
if missing_features:
    print(f"Các trường bị thiếu: {missing_features}")
else:
    print("Tất cả các trường đều tồn tại trong dữ liệu!")

# Trích xuất dữ liệu training
X_train = train_values[binary_features].copy()
y_train = train_labels['damage_grade'].copy()

# Trích xuất dữ liệu test
X_test = test_values[binary_features].copy()

print(f"\nShape của X_train: {X_train.shape}")
print(f"Shape của y_train: {y_train.shape}")
print(f"Shape của X_test: {X_test.shape}")

# Kiểm tra dữ liệu
print(f"\nKiểm tra giá trị unique trong các trường nhị phân:")
for feature in binary_features[:5]:  # Hiển thị 5 cột đầu
    unique_vals = X_train[feature].unique()
    print(f"{feature}: {unique_vals}")
    
print("\nMẫu dữ liệu X_train:")
print(X_train.head())
```

```python
target_feature = 'damage_grade'

# Bước 1: Tính các xác suất p_k := N_k / N cho mỗi class
print("Bước 1: Tính xác suất prior cho mỗi class...")

Y = y_train  # damage_grade từ training data
labels = np.zeros((len(set(Y)), 2), dtype=object) 

# Liệt kê các nhãn (label) khác nhau và tính xác suất P_k của mỗi nhãn
id = 0
for label in sorted(set(Y)):  # sắp xếp để đảm bảo thứ tự 1, 2, 3
    labels[id, 0] = label
    labels[id, 1] = (Y == label).sum() / len(Y)
    print(f"Class {label}: {(Y == label).sum()} samples, probability = {labels[id, 1]:.4f}")
    id += 1

print(f"\nLabels array shape: {labels.shape}")
print("Labels array:")
print(labels)
```

```python
# Bước 2: Định nghĩa hàm tính P_ik - xác suất feature i = 1 trong class k
def Pik_feature_per_class(X):
    """
    Tính xác suất feature nhận giá trị 1 trong một class
    Returns the prob. of feature per class
    """
    X = np.array(X)
    # count the number of element x_m which has field i^th : x_im = 1
    count = (X == 1).sum()
    # P_ik := [|{xm in class k : x_im = 1}|] / N_k 
    return count / len(X) if len(X) > 0 else 0

print("Hàm Pik_feature_per_class đã được định nghĩa!")
```

```python
# Bước 3: Định nghĩa hàm dự đoán
def predict_output_label(X, x_input, p_labels, target_feature):
    """
    Dự đoán nhãn cho một mẫu dữ liệu đầu vào
    
    Parameters:
    X: DataFrame chứa dữ liệu training
    x_input: Series chứa dữ liệu cần dự đoán (chỉ các features, không có target)
    p_labels: array chứa các class và xác suất prior
    target_feature: tên của cột target
    """
    # Score for each class (bắt đầu với log prior)
    p = np.log(np.array(p_labels[:, 1], dtype=float))

    for k in range(len(p)):
        # Lấy tất cả mẫu thuộc class k
        Xk = X[X[target_feature] == p_labels[k, 0]]
        
        # Tính likelihood cho từng feature
        for i in range(len(x_input)):
            # Tính P_ik: xác suất feature i = 1 trong class k
            Pik = Pik_feature_per_class(Xk.iloc[:, i])  # Chú ý: i tương ứng với cột trong X_train
            
            if Pik != 0 and Pik != 1:  # Tránh log(0)
                # Bernoulli likelihood: P(x_i | class_k)
                # Nếu x_i = 1: P_ik, nếu x_i = 0: (1 - P_ik)
                likelihood = x_input.iloc[i] * Pik + (1 - x_input.iloc[i]) * (1 - Pik)
                p[k] += np.log(likelihood)
            elif Pik == 0:  # Tất cả mẫu trong class k có feature i = 0
                if x_input.iloc[i] == 1:
                    p[k] += np.log(1e-10)  # Xác suất rất nhỏ thay vì 0
                # Nếu x_input.iloc[i] == 0 thì likelihood = 1, log(1) = 0
            elif Pik == 1:  # Tất cả mẫu trong class k có feature i = 1
                if x_input.iloc[i] == 0:
                    p[k] += np.log(1e-10)  # Xác suất rất nhỏ thay vì 0
                # Nếu x_input.iloc[i] == 1 thì likelihood = 1, log(1) = 0

    # Chọn class có score cao nhất
    y_star = np.argmax(p)
    return p_labels[y_star, 0]

print("Hàm predict_output_label đã được định nghĩa!")
```

```python
print("Bước 4: Chuẩn bị dữ liệu training...")

# Tạo DataFrame training bao gồm cả features và target
X_Train_combined = X_train.copy()
X_Train_combined[target_feature] = y_train

print(f"X_Train_combined shape: {X_Train_combined.shape}")
print(f"Columns: {list(X_Train_combined.columns)}")
print(f"Target feature: {target_feature}")

# Kiểm tra mẫu dữ liệu
print("\nMẫu dữ liệu X_Train_combined:")
print(X_Train_combined.head())
```

```python
# Bước 5: Thực hiện dự đoán trên tập test (test với subset nhỏ trước)
print("Bước 5: Dự đoán trên tập test...")

# Test với 100 mẫu đầu tiên để kiểm tra
test_size = 100
print(f"Testing với {test_size} mẫu đầu tiên...")

X_test_small = X_test.iloc[:test_size].copy()
Y_pred_small = np.zeros(test_size, dtype=y_train.dtype)

# Dự đoán cho từng mẫu trong test set nhỏ
print("Đang dự đoán...")
for i in range(test_size):
    if i % 25 == 0:  # In tiến trình mỗi 25 mẫu
        print(f"Đã xử lý {i}/{test_size} mẫu...")
    
    # Dự đoán cho mẫu thứ i
    Y_pred_small[i] = predict_output_label(X_Train_combined, X_test_small.iloc[i, :], labels, target_feature)

print(f"Hoàn thành dự đoán cho {test_size} mẫu!")
print(f"Kết quả dự đoán: {Y_pred_small}")

# Phân phối kết quả dự đoán
unique, counts = np.unique(Y_pred_small, return_counts=True)
print(f"\nPhân phối dự đoán (trên {test_size} mẫu):")
for class_val, count in zip(unique, counts):
    print(f"Class {class_val}: {count} mẫu ({count/test_size*100:.2f}%)")

# Test thử với 1 mẫu cụ thể để kiểm tra chi tiết
print(f"\nTest với mẫu đầu tiên:")
print(f"Input features: {X_test_small.iloc[0].values}")
print(f"Predicted class: {Y_pred_small[0]}")
```
