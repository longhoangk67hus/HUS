## Bài tập 1:
### Lớp bài toán *P* (Polynomial time): là tập hợp các bài toán quyết định (decision problems) có thể giải bằng thuật toán trong thời gian đa thức (polynomial time), tức là có một thuật toán chạy trong thời gian $O(n^k)$ với `k` là hằng số.

  **Ví dụ**: Kiểm tra số nguyên tố, sắp xếp dãy số.

### Lớp bài toán NP: (Nondeterministic Polynomial time) là tập hợp các bài toán mà lời giải có thể **kiểm tra** được trong thời gian đa thức, tức là nếu ai đó đưa ra lời giải thì ta có thể xác minh đúng/sai một cách nhanh chóng.

  **Ví dụ**: Bài toán người du lịch (TSP), bài toán thỏa mãn logic (SAT).
### Lớp bài toán Co-NP (Complement of NP): Là tập hợp các bài toán có bài toán bổ sung thuộc lớp NP. Tức là nếu một bài toán A thuộc Co-NP, thì phủ định của nó (not A) thuộc NP.
  
  **Ví dụ**: Nếu bài toán “có một nghiệm thỏa mãn công thức logic?” thuộc NP, thì bài toán “mọi nghiệm đều không thỏa?” thuộc Co-NP.
- Mối liên hệ giữa P, NP và Co-NP:
  - Rõ ràng: $$\text{P} \subseteq \text{NP} \quad \text{và} \quad \text{P} \subseteq \text{Co-NP}$$
  - Tuy nhiên, chưa biết liệu $$\text{NP} = \text{Co-NP} $$
  - Nếu:
   $$ \text{P} = \text{NP} $$
  thì:
  $$\text{NP} = \text{Co-NP} = \text{P}$$
### Lớp bài toán NPC (NP- complete): Là giao giữa `NP` và `NP-hard`, tức là:
- Nằm trong `NP`
- Mọi bài toán trong `NP` đều có thể giảm về nó trong thời gian đa thức.

**Ví dụ**: Bài toán `3-SAT`, bài toán `Maximum Independent Set`.

### Mối liên hệ giữa P, NP, Co-NP, NP-Hard và NPC:
- $ \text{P} \subseteq \text{NP} \subseteq \text{NP-Hard} $
- $ \text{NPC} = \text{NP} \cap \text{NP-Hard} $
- $ \text{P} \subseteq \text{Co-NP}$, nhưng chưa biết liệu $ \text{NP} = \text{Co-NP} $
- Nếu tìm được một bài toán `NP-complete` thuộc `P`, thì mọi bài toán trong `NP` đều thuộc `P` → $ \text{P} = \text{NP} $

### N = NP?
Đây là một trong những câu hỏi mở lớn nhất trong khoa học máy tính.
- Nếu $\text{P} = \text{NP}$: Mọi bài toán có lời giải kiểm chứng được thì cũng có thể giải được nhanh.
- Nếu $\text{P} \ne \text{NP}$: Có những bài toán rất khó để giải, dù có thể dễ dàng kiểm tra lời giải.


## Bài 2:
**Theo bạn, như thế nào là một thuật toán nhanh? Thuật toán hiệu quả?  Độ phức tạp của thuật toán?**

- Một thuật toán **nhanh** là thuật toán có độ phức tạp thời gian **đa thức**, tức là thời gian thực hiện tăng không quá nhanh khi kích thước đầu vào tăng (ví dụ: $O(n)$, $O(n log n)$, $O(n²)$).
- Thuật toán **hiệu quả** thường được hiểu là có thể chạy được trong thời gian chấp nhận được trên thực tế và thường thuộc lớp `P`.
- **Độ phức tạp của thuật toán** là cách để đo lường lượng tài nguyên (thời gian, bộ nhớ) cần dùng, được biểu diễn bằng ký hiệu $Big-O$ như $O(n^2)$, $O(2^n)$.

**Thế nào là bài toán dễ và khó?**
- "Dễ": Có thuật toán giải trong thời gian đa thức (`P`).
- "Khó": Không biết thuật toán đa thức, thường là các bài toán `NP-hard`, `NPC`.


### Thuật toán dễ: **QuickSort**
#### **Bài toán:** Sắp xếp một mảng số nguyên theo thứ tự tăng dần.
#### **Thiết kế thuật toán:**
1. **Chọn pivot** (chốt): chọn phần tử giữa, đầu hoặc ngẫu nhiên trong mảng.
2. **Chia mảng**: đưa các phần tử nhỏ hơn pivot về bên trái, lớn hơn pivot về bên phải.
3. **Đệ quy**: áp dụng tiếp tục cho hai mảng con.

```python
def quicksort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr)//2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quicksort(left) + middle + quicksort(right)
```
### Phân tích độ phức tạp:
- **Trung bình**: $O(nlogn)$
- **Tệ nhất:** $O(n^2)$ (khi pivot chọn không tốt)
### Đánh giá:
- Hiệu quả và dễ cài đặt
- Được dùng thực tế trong thư viện chuẩn của nhiều ngôn ngữ.
- Độ phức tạp đa thức → nằm trong lớp `P`

### Thuật toán khó: **SAT (Boolean Satisfiability Problem)**
#### **Bài toán:** Cho một biểu thức logic ở dạng CNF (Chuẩn tắc hội tụ - conjunction of disjunctions), tìm xem có tồn tại gán giá trị TRUE/FALSE cho các biến sao cho biểu thức đúng hay không?
**Ví dụ:**
```text
(x1 ∨ ¬x2 ∨ x3) ∧ (¬x1 ∨ x2) ∧ (¬x3)
```
#### ** Thiết kế thuật toán (DPLL - Davis-Putnam-Logemann-Loveland):**
**1**. **Đơn giản biểu thức**: Loại bỏ mệnh đề đã thỏa mãn.

**2**. **Gán thử giá trị** cho một biến.

**3**. **Đệ quy** kiểm tra biểu thức còn lại.

**4**. **Backtrack** nếu không hợp lệ.
```python
def DPLL(clauses, assignment):
    if all_true(clauses, assignment):
        return True
    if has_contradiction(clauses, assignment):
        return False
    variable = choose_unassigned_variable(clauses, assignment)
    for value in [True, False]:
        assignment[variable] = value
        if DPLL(clauses, assignment):
            return True
        del assignment[variable]
    return False
```
#### **Phân tích độ phức tạp:**
- Trường hợp xấu nhất: $O(2^n)$, với n là số biến.
- Dù có cải tiến (DPLL, CDCL), độ phức tạp vẫn là **hàm mũ** trong trường hợp tổng quát.
#### **Đánh giá:**
- Là bài toán **NP-complete**
- Không có thuật toán đa thức đã biết
- Thuộc lớp bài toán "khó"



### **Bài 3: Có thể giải quyết tất cả các bài toán trong thời gian đa thức không? Vì sao?**
Không chắc chắn. Câu hỏi này liên quan đến bài toán mở **P = NP?**

- Nếu **P = NP** → tất cả bài toán trong `NP` đều có thể giải trong thời gian đa thức.
- Nếu **P ≠ NP** → tồn tại bài toán trong `NP` nhưng không giải được trong thời gian đa thức.

Hiện tại chưa có chứng minh nào cho cả hai phía.

## Bài tập 4: Có phải bất kỳ bài toán nào cũng là NP-Hard không? Vì sao?
Không. Không phải mọi bài toán đều là `NP-hard`.

- `NP-hard` là lớp các bài toán khó **ít nhất** như mọi bài toán trong `NP`.
- Các bài toán đơn giản như cộng hai số, kiểm tra số chẵn/lẻ không phải là `NP-hard`.

Chỉ những bài toán có thể **mã hóa lại các bài toán trong NP** một cách đa thức mới được coi là `NP-hard`.


## Bài tập 5: Lớp bài toán NP có phải là lớp bài toán không thể giải được trong thời gian đa thức hay không? Vì sao?
Không nhất thiết.

- `NP` là lớp các bài toán mà lời giải có thể **kiểm tra** trong thời gian đa thức.
- Có thể một số bài toán trong `NP` vẫn thuộc `P` → vẫn giải được trong thời gian đa thức.
- Việc có **giải được tất cả bài toán NP trong đa thức hay không** là nội dung của bài toán **P = NP?**

## Bài tập 6: Nếu $ \text{P} = \text{NP} $ thì $ \text{NPC} \subseteq \text{P} $. Theo bạn, đúng hay sai và vì sao?
**Đúng.**

- `NPC` (NP-complete) là tập hợp các bài toán khó nhất trong `NP`.
- Nếu `P = NP` thì mọi bài toán trong `NP`, bao gồm cả `NPC`, đều thuộc `P`.
→  $\text{NPC} \subseteq \text{P} $

## Bài tập 7

**Nếu $ \text{NPC} \subseteq \text{P} $ thì $ \text{P} = \text{NP} $. Theo bạn, đúng hay sai và vì sao?**

 **Đúng.**

- Nếu một bài toán `NP-complete` có thuật toán trong `P`, thì mọi bài toán trong `NP` đều có thể **giảm về** bài toán đó trong thời gian đa thức.
- Vì vậy, mọi bài toán trong `NP` cũng thuộc `P`.

→ Kết luận: $ \text{NPC} \subseteq \text{P} \Rightarrow \text{P} = \text{NP} $
